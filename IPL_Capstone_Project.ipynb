# Loading Libraries and Dataset

# Dataset Analysing
1. Which team won the most matches?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y =match_wins.index, x = match_wins.values, palette = 'viridis')
plt.title("Most Match Winning Teams")
plt.xlabel("No. of Wins")
plt.ylabel("Teams")
```

```python
sns.countplot(x = df['toss_decision'], palette = 'rainbow')
plt.title("Toss Decision Trends")
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count *100)/df.shape[0]
print(percentage.round(2))
```

```python
df.head()
```

```python
sns.countplot(x = df['won_by'], palette = 'rainbow')
plt.title("Won_By")
```

# Key Player Performance

1.Most "Player of the match" awards.

```python
count = df["player_of_the_match"].value_counts().head(10)
print(count)
sns.barplot(y =count.index, x = count.values, palette = 'mako')
plt.title("Top 10 Player of the Match")
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)   # 
high
```

```python
high.plot(kind = 'barh')
```

```python
df['highest_wicket'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['highest_wicket'] = df['highest_wicket'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wicket'].sum().sort_values(ascending=False).head(10)
top_bowlers
```

```python
top_bowlers.plot(kind = 'barh')
```
